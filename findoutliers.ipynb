```python
# IQR method
import pandas as pd;
import matplotlib.pyplot as plt;
traind_df = pd.read_csv("https://download.mlcc.google.com/mledu-datasets/california_housing_train.csv");
Q1 = traind_df["median_house_value"].quantile(0.25);
Q3 = traind_df["median_house_value"].quantile(0.75);
IQR = Q3 - Q1
lower_bound = (Q1 - 1.5) *IQR
upper_bound = (Q3 + 1.5) *IQR
outliers = traind_df[(traind_df["median_house_value"] < lower_bound) | (traind_df["median_house_value"] > upper_bound)]
print(outliers)
```
